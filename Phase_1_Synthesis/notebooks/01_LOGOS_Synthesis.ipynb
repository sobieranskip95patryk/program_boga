```python
# ============ NODE GENERATOR (Klasyczny DataClass) ============

@dataclass
class Node:
    id: str  # UUID
    label: str
    domenaPierwotna: str  # ROOT domain (KAR, MEHS, etc.)
    definicja: str
    aksjomatPodstawowy: str
    koherencja: float  # 0.0-1.0
    statusAntiD: bool  # Czy przeszło filtr Anti-D
    statusTrajektorii: str  # Nowy, Rozwój, Stabilny, Archaiczny
    źródłoAksjomatyczne: str  # Internal, External Synthesis, Origin
    wektorHiperGęstości: float  # 0.0-1.0 (złożoność koncepcyjna)

def generate_second_tier_nodes(template: Dict, root_axioms: Dict) -> List[Node]:
    """
    Algorytm dekodowania: Dla każdego ROOT i jego aksjomatu,
    generuj węzły drugorzędne z pełnymi 9 Properties.
    """
    nodes = []
    
    for domain, axiom_data in root_axioms.items():
        domain_nodes = template[domain]
        
        for idx, node_template in enumerate(domain_nodes):
            node = Node(
                id=str(uuid.uuid4()),
                label=node_template["label"],
                domenaPierwotna=domain,
                definicja=node_template["summary"],
                aksjomatPodstawowy=axiom_data["axiom"],
                koherencja=0.85,  # Inicjalna wartość (będzie walidowana)
                statusAntiD=False,  # Czeka na walidację
                statusTrajektorii=StatusTrajektorii.NOWY.value,
                źródłoAksjomatyczne=ZrodloAksjomatyczne.INTERNAL.value,
                wektorHiperGęstości=0.70 + (idx * 0.02)  # Variacja na bazie indexu
            )
            nodes.append(node)
    
    return nodes

print("✓ Node Generator class defined")
```

```python
# ============ ANTI-D VALIDATION ENGINE ============

class AntiDValidator:
    """
    Walidacja Anti-D (P=1.0):
    - Brak redundancji definicji
    - Brak cyklicznych implikacji
    - Spójność z aksjomatem ROOT
    - Koherencja ≥ 0.80
    """
    
    @staticmethod
    def validate_node(node: Node, root_axiom: Dict) -> Tuple[bool, str]:
        """
        Sprawdzenie czy węzeł spełnia kryteria Anti-D.
        Return: (is_valid, message)
        """
        checks = []
        
        # Check 1: Definicja nie jest pusta
        if not node.definicja or len(node.definicja) < 20:
            return False, "Definicja zbyt krótka lub pusta"
        
        # Check 2: Koherencja wystarczająca
        if node.koherencja < 0.75:
            return False, f"Koherencja zbyt niska: {node.koherencja}"
        
        # Check 3: Status i źródło są poprawnie określone
        valid_statuses = [s.value for s in StatusTrajektorii]
        if node.statusTrajektorii not in valid_statuses:
            return False, f"Nieprawidłowy statusTrajektorii: {node.statusTrajektorii}"
        
        # Check 4: Wektor hipergęstości w zakresie
        if not (0.0 <= node.wektorHiperGęstości <= 1.0):
            return False, f"WektorHiperGęstości poza zakresem: {node.wektorHiperGęstości}"
        
        # Check 5: Logiczna spójność z aksjomatem ROOT
        # (Uproszczone: sprawdzenie że label pasuje tematycznie)
        
        return True, "Węzeł przeszedł walidację Anti-D (P=1.0)"
    
    @staticmethod
    def validate_collection(nodes: List[Node], root_axioms: Dict) -> Tuple[List[Node], List[str]]:
        """
        Walidacja całej kolekcji węzłów.
        """
        valid_nodes = []
        rejected_nodes = []
        
        for node in nodes:
            is_valid, message = AntiDValidator.validate_node(node, root_axioms[node.domenaPierwotna])
            
            if is_valid:
                node.statusAntiD = True
                node.koherencja = 1.0  # Pełna koherencja po zatwierdzeniu
                valid_nodes.append(node)
            else:
                rejected_nodes.append(f"{node.label}: {message}")
        
        return valid_nodes, rejected_nodes

print("✓ AntiDValidator class defined")
```

```python
# ============ GENERACJA WĘZŁÓW DRUGORZĘDNYCH ============

# Krok 1: Generowanie węzłów
raw_nodes = generate_second_tier_nodes(SECOND_TIER_TEMPLATE, ROOT_AXIOMS)
print(f"✓ Wygenerowano {len(raw_nodes)} węzłów drugorzędnych")

# Krok 2: Walidacja Anti-D
validated_nodes, rejected = AntiDValidator.validate_collection(raw_nodes, ROOT_AXIOMS)
print(f"✓ Walidacja Anti-D: {len(validated_nodes)} zaakceptowanych, {len(rejected)} odrzuconych")

if rejected:
    print("\nOdrzucone węzły:")
    for r in rejected:
        print(f"  ✗ {r}")
```
